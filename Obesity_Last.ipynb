  <h2 style="margin:6px 0 8px 0; color:#12343B; font-size:25px; font-weight:850;">Why the Target Classes Were Grouped</h2>
  <p style="margin:0 0 10px 0; color:#48636B; line-height:1.65; font-size:15px;">
    The original target contains seven detailed labels: Insufficient Weight, Normal Weight, Overweight Level I, Overweight Level II, Obesity Type I, Obesity Type II, and Obesity Type III. In this notebook, these labels are grouped into four broader and more interpretable classes: Underweight, Normal, Overweight, and Obesity.
  </p>
  <ul style="margin: 8px 0 0 20px; color:#27464D; line-height:1.7; font-size:15px;">
    <li><b>Modeling clarity:</b> The grouped classes reduce unnecessary target fragmentation and make the classification task easier to interpret.</li>
    <li><b>Practical interpretation:</b> The difference between obesity subtypes is less important for the main machine learning objective, because all obesity categories indicate a high risk group.</li>
    <li><b>Real world meaning:</b> Once a person falls into the obesity group, the next practical step is clinical assessment or referral to a healthcare professional, rather than treating each obesity subtype as a completely separate machine learning decision.</li>
    <li><b>Evaluation stability:</b> Grouping also helps reduce class sparsity and supports more reliable evaluation using Macro F1.</li>
  </ul>
  <div style="margin-top:12px; padding:10px 12px; border-radius:12px; background:#EAF6FF; color:#24454D; line-height:1.6;">
    <b style="color:#0077B6;">Final target design:</b> Seven original labels are converted into four clinically meaningful groups to support clearer prediction, interpretation, and reporting.
  </div>

```python
# =========================
# OUTLIER DETECTION
# =========================

outlier_summary = []

for col in numerical_cols:
    Q1 = train[col].quantile(0.25)
    Q3 = train[col].quantile(0.75)
    IQR = Q3 - Q1
    
    if IQR == 0:
        outlier_summary.append([
            col,
            "Skipped",
            "IQR = 0",
            round(Q1, 2),
            round(Q3, 2)
        ])
        continue
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = ((train[col] < lower_bound) | (train[col] > upper_bound)).sum()
    outlier_percentage = (outlier_count / len(train)) * 100
    
    outlier_summary.append([
        col,
        outlier_count,
        round(outlier_percentage, 2),
        round(lower_bound, 2),
        round(upper_bound, 2)
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=["Feature", "Outlier Count", "Outlier Percentage", "Lower Bound", "Upper Bound"]
)

display(outlier_df)
```

  <ul style="margin: 10px 0 0 20px; color:#27464D; line-height:1.65;"><li><b>ANOVA F test</b> checks whether numerical feature means differ across target classes.</li><li><b>Eta squared</b> measures numerical feature association strength with the target.</li><li>Larger eta squared values indicate stronger separation between target groups.</li></ul>

```python
# =========================
# CORRELATION SECTION
# Numerical Features & Target
# =========================

num_target_results = []

for col in numerical_cols:
    
    groups = [
        train.loc[train[target_col] == cls, col].dropna()
        for cls in train[target_col].unique()
    ]
    
    f_stat, p_value = f_oneway(*groups)
    
    overall_mean = train[col].mean()
    
    ss_between = sum(
        len(group) * (group.mean() - overall_mean) ** 2
        for group in groups
    )
    
    ss_total = sum(
        (train[col].dropna() - overall_mean) ** 2
    )
    
    eta_squared = ss_between / ss_total if ss_total != 0 else 0
    
    num_target_results.append([
        col,
        f_stat,
        p_value,
        eta_squared
    ])

num_target_df = pd.DataFrame(
    num_target_results,
    columns=[
        "Numerical Feature",
        "F Statistic",
        "P-value",
        "Eta Squared"
    ]
)

num_target_df = (
    num_target_df
    .sort_values("Eta Squared", ascending=False)
    .reset_index(drop=True)
)

display(num_target_df.round(4))


plt.figure(figsize=(10, 5))

num_bar_colors = (PALETTE_BLUE_GREEN_RED * ((len(num_target_df) // len(PALETTE_BLUE_GREEN_RED)) + 1))[:len(num_target_df)]

plt.bar(
    num_target_df["Numerical Feature"],
    num_target_df["Eta Squared"],
    color=num_bar_colors,
    edgecolor=DARK_BLUE
)

plt.title("Numerical Features Association with Target")
plt.xlabel("Numerical Feature")
plt.ylabel("Eta Squared")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

  <p style="margin:0; color:#48636B; line-height:1.55;">Evaluate categorical feature association with the grouped target using Chi square testing and corrected Cramer's V.</p>
  <ul style="margin: 8px 0 0 18px; color:#27464D; line-height:1.6;"><li>Chi square checks statistical association.</li><li>Corrected Cramer's V measures association strength.</li></ul>
</div>


```python
# =========================
# CORRELATION SECTION
# Categorical Features & Target
# =========================

def cramers_v(x, y):
    
    confusion_table = pd.crosstab(x, y)
    
    chi2 = chi2_contingency(confusion_table, correction=False)[0]
    n = confusion_table.sum().sum()
    
    rows, cols = confusion_table.shape
    
    if n <= 1 or rows <= 1 or cols <= 1:
        return 0
    
    phi2 = chi2 / n
    
    # Bias-corrected Cramer's V
    phi2_corrected = max(
        0,
        phi2 - ((cols - 1) * (rows - 1)) / (n - 1)
    )
    
    rows_corrected = rows - ((rows - 1) ** 2) / (n - 1)
    cols_corrected = cols - ((cols - 1) ** 2) / (n - 1)
    
    denominator = min(rows_corrected - 1, cols_corrected - 1)
    
    if denominator <= 0:
        return 0
    
    return np.sqrt(phi2_corrected / denominator)


cat_target_results = []

for col in categorical_cols:
    
    table = pd.crosstab(train[col], train[target_col])
    
    chi2_stat, p_value, dof, expected = chi2_contingency(table)
    
    cv_value = cramers_v(train[col], train[target_col])
    
    cat_target_results.append([
        col,
        chi2_stat,
        p_value,
        cv_value
    ])

cat_target_df = pd.DataFrame(
    cat_target_results,
    columns=[
        "Categorical Feature",
        "Chi-square Statistic",
        "P-value",
        "Corrected Cramer's V"
    ]
)

cat_target_df = (
    cat_target_df
    .sort_values("Corrected Cramer's V", ascending=False)
    .reset_index(drop=True)
)

display(cat_target_df.round(4))


plt.figure(figsize=(10, 5))

cat_bar_colors = (PALETTE_BLUE_GREEN_RED * ((len(cat_target_df) // len(PALETTE_BLUE_GREEN_RED)) + 1))[:len(cat_target_df)]

plt.bar(
    cat_target_df["Categorical Feature"],
    cat_target_df["Corrected Cramer's V"],
    color=cat_bar_colors,
    edgecolor=DARK_BLUE
)

plt.title("Categorical Features Association with Target")
plt.xlabel("Categorical Feature")
plt.ylabel("Corrected Cramer's V")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# =========================
# MULTICOLLINEARITY CHECKER
# Numerical Features with Numerical Features
# =========================

num_corr_matrix = train[numerical_cols].corr(method="spearman")

display(num_corr_matrix.round(3))


plt.figure(figsize=(10, 7))

plt.imshow(num_corr_matrix, cmap=HEAT_CMAP)
plt.colorbar()

plt.xticks(
    range(len(numerical_cols)),
    numerical_cols,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(numerical_cols)),
    numerical_cols
)

plt.title("Numerical Features Correlation Matrix")

for i in range(len(numerical_cols)):
    for j in range(len(numerical_cols)):
        plt.text(
            j,
            i,
            round(num_corr_matrix.iloc[i, j], 2),
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()


num_corr_pairs = (
    num_corr_matrix
    .abs()
    .unstack()
    .reset_index()
)

num_corr_pairs.columns = [
    "Feature 1",
    "Feature 2",
    "Correlation"
]

num_corr_pairs = num_corr_pairs[
    num_corr_pairs["Feature 1"] != num_corr_pairs["Feature 2"]
]

num_corr_pairs["Pair"] = num_corr_pairs.apply(
    lambda row: tuple(sorted([row["Feature 1"], row["Feature 2"]])),
    axis=1
)

num_corr_pairs = (
    num_corr_pairs
    .drop_duplicates("Pair")
    .drop(columns="Pair")
    .sort_values("Correlation", ascending=False)
    .reset_index(drop=True)
)

print("Highly correlated numerical feature pairs:")
display(num_corr_pairs[num_corr_pairs["Correlation"] >= 0.80])
```

```python
# =========================
# MULTICOLLINEARITY CHECKER
# Numerical Features with Categorical Features
# =========================

def correlation_ratio(categories, measurements):
    
    data = pd.DataFrame({
        "category": categories,
        "measurement": measurements
    }).dropna()
    
    if data["category"].nunique() <= 1:
        return 0
    
    overall_mean = data["measurement"].mean()
    
    ss_between = 0
    ss_total = sum(
        (data["measurement"] - overall_mean) ** 2
    )
    
    for category in data["category"].unique():
        group = data.loc[
            data["category"] == category,
            "measurement"
        ]
        
        ss_between += len(group) * (group.mean() - overall_mean) ** 2
    
    if ss_total == 0:
        return 0
    
    eta_squared = ss_between / ss_total
    
    return np.sqrt(eta_squared)


num_cat_results = []

for num_col in numerical_cols:
    for cat_col in categorical_cols:
        
        eta_value = correlation_ratio(
            train[cat_col],
            train[num_col]
        )
        
        num_cat_results.append([
            num_col,
            cat_col,
            eta_value
        ])

num_cat_assoc_df = pd.DataFrame(
    num_cat_results,
    columns=[
        "Numerical Feature",
        "Categorical Feature",
        "Correlation Ratio"
    ]
)

num_cat_assoc_df = (
    num_cat_assoc_df
    .sort_values("Correlation Ratio", ascending=False)
    .reset_index(drop=True)
)

display(num_cat_assoc_df.round(4))


print("High numerical-categorical associations:")
display(
    num_cat_assoc_df[
        num_cat_assoc_df["Correlation Ratio"] >= 0.70
    ]
)
```

  <ul style="margin: 10px 0 0 20px; color:#27464D; line-height:1.65;"><li>Select final input features.</li><li>Separate X from y.</li><li>Split the data using stratification.</li><li>Compute class and sample weights for imbalance.</li><li>Build the preprocessing pipeline.</li></ul>
</div>


```python
# =========================
# DATA PREPARATION
# =========================

final_numerical_cols = numerical_cols.copy()
final_categorical_cols = categorical_cols.copy()

final_features = final_numerical_cols + final_categorical_cols

print("Final numerical columns:")
print(final_numerical_cols)

print("\nFinal categorical columns:")
print(final_categorical_cols)

print("\nNumber of final features:", len(final_features))
```

```python
X = train[final_features]
y = train[target_col]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# =========================
# TRAIN / VALIDATION / TEST SPLIT
# =========================

# The data is split into three parts:
# - Training set: used to fit models.
# - Validation set: used for model comparison, tuning decisions, BMI label-proxy checks, and Top 10 feature experiment.
# - Test set: reserved for the final selected model evaluation only.

X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.25,
    random_state=42,
    stratify=y_train_val
)

print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_val shape:", y_val.shape)
print("y_test shape:", y_test.shape)
```

```python
# =========================
# SAMPLE WEIGHTS FOR IMBALANCED DATA
# =========================

class_counts = y_train.value_counts()

train_sample_weight = y_train.map(
    lambda cls: len(y_train) / (len(class_counts) * class_counts[cls])
).values

display(class_counts.to_frame("count"))
```

The data is split into training, validation, and test sets using stratification. The validation set is used for model selection and development checks, while the test set is reserved for the final selected model evaluation only. Sample weights are computed from the training set to handle class imbalance consistently during model fitting.</span>

  <h3 style="margin: 4px 0; color:#12343B; font-size:21px; font-weight:800;">Preprocessing Pipeline</h3>
  <p style="margin:0; color:#48636B; line-height:1.55;">Apply different preprocessing steps based on feature type so the model receives clean and consistent inputs.</p>
  <ul style="margin: 8px 0 0 18px; color:#27464D; line-height:1.6;"><li>Numerical features are scaled using <b>RobustScaler</b>.</li><li>Ordinal categorical features are encoded using <b>OrdinalEncoder</b>.</li><li>Nominal categorical features are encoded using <b>OneHotEncoder</b>.</li></ul>
</div>


```python
# =========================
# PREPROCESSING PIPELINE
# Encoding + scaling
# =========================

ordinal_cols = [
    "CAEC",
    "CALC"
]

nominal_cols = [
    col for col in final_categorical_cols
    if col not in ordinal_cols
]

# CAEC and CALC are ordinal because their categories have a natural order.
# CALC "Always" does not appear in train.csv, but it appears in test.csv,
# so it is kept here to avoid treating it as an unknown category later.
ordinal_categories = [
    ["no", "Sometimes", "Frequently", "Always"],  # CAEC
    ["no", "Sometimes", "Frequently", "Always"]   # CALC
]

numeric_transformer = Pipeline(steps=[
    ("scaler", RobustScaler())
])

ordinal_transformer = Pipeline(steps=[
    ("encoder", OrdinalEncoder(
        categories=ordinal_categories,
        handle_unknown="use_encoded_value",
        unknown_value=-1
    ))
])

nominal_transformer = Pipeline(steps=[
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, final_numerical_cols),
        ("ord", ordinal_transformer, ordinal_cols),
        ("nom", nominal_transformer, nominal_cols)
    ]
)
```

  <span style="color:#24454D; line-height:1.65;">The preprocessing pipeline is appropriate because it treats numerical, ordinal, and nominal variables differently. <code>CAEC</code> and <code>CALC</code> are handled as ordinal features because their categories have a meaningful order. Other categorical variables are one hot encoded because they do not have a natural numerical order.</span>

```python
# =========================
# MODEL COMPARISON
# Main models before hyperparameter tuning
# Validation set is used for model comparison.
# =========================

target_encoder = LabelEncoder()

y_train_encoded = target_encoder.fit_transform(y_train)
y_val_encoded = target_encoder.transform(y_val)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),

    "SVC": SVC(
        kernel="rbf",
        C=1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=600,
        max_depth=4,
        learning_rate=0.03,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="multi:softprob",
        eval_metric="mlogloss",
        random_state=42,
        n_jobs=-1
    )
}

results = []
fitted_models = {}

for name, model in models.items():

    clf = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    if name == "XGBoost":
        clf.fit(
            X_train,
            y_train_encoded,
            model__sample_weight=train_sample_weight
        )
        y_pred_encoded = clf.predict(X_val)
        y_pred_model = target_encoder.inverse_transform(y_pred_encoded)

    else:
        clf.fit(
            X_train,
            y_train,
            model__sample_weight=train_sample_weight
        )
        y_pred_model = clf.predict(X_val)

    acc = accuracy_score(y_val, y_pred_model)
    f1 = f1_score(y_val, y_pred_model, average="macro")

    results.append([name, acc, f1])
    fitted_models[name] = clf

results_df = pd.DataFrame(
    results,
    columns=["Model", "Validation Accuracy", "Validation Macro F1"]
)

results_df = (
    results_df
    .sort_values("Validation Macro F1", ascending=False)
    .reset_index(drop=True)
)

display(results_df.round(4))
```

```python
# =========================
# RANDOM FOREST TUNING
# =========================

rf_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ))
])

rf_param_grid = {
    "model__n_estimators": [300, 500, 700],
    "model__max_depth": [None, 10, 15, 20],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__max_features": ["sqrt", "log2", None],
    "model__bootstrap": [True]
}

rf_search = RandomizedSearchCV(
    estimator=rf_pipe,
    param_distributions=rf_param_grid,
    n_iter=30,
    scoring="f1_macro",
    cv=5,
    random_state=42,
    n_jobs=-1
)

rf_search.fit(
    X_train,
    y_train,
    model__sample_weight=train_sample_weight
)

best_rf_model = rf_search.best_estimator_

y_pred_rf_tuned = best_rf_model.predict(X_val)

print("Best Parameters:")
print(rf_search.best_params_)

print("\nBest CV Macro F1:", rf_search.best_score_)
print("Validation Accuracy:", accuracy_score(y_val, y_pred_rf_tuned))
print("Validation Macro F1:", f1_score(y_val, y_pred_rf_tuned, average="macro"))

print("\nValidation Classification Report:")
print(classification_report(y_val, y_pred_rf_tuned))

```

```python
# =========================
# XGBOOST TUNING
# =========================

xgb_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", XGBClassifier(
        objective="multi:softprob",
        eval_metric="mlogloss",
        device="cuda"
        random_state=42,
        n_jobs=-1
    ))
])

param_grid = {
    "model__n_estimators": [400, 600, 800],
    "model__max_depth": [3, 4, 5],
    "model__learning_rate": [0.02, 0.03, 0.05],
    "model__subsample": [0.8, 0.9, 1.0],
    "model__colsample_bytree": [0.8, 0.9, 1.0],
    "model__min_child_weight": [1, 3, 5],
    "model__gamma": [0, 0.1, 0.3],
    "model__reg_lambda": [1, 3, 5]
}

xgb_search = RandomizedSearchCV(
    estimator=xgb_pipe,
    param_distributions=param_grid,
    n_iter=30,
    scoring="f1_macro",
    cv=5,
    random_state=42,
    n_jobs=-1
)

xgb_search.fit(
    X_train,
    y_train_encoded,
    model__sample_weight=train_sample_weight
)

best_xgb_model = xgb_search.best_estimator_

y_pred_encoded = best_xgb_model.predict(X_val)
y_pred_xgb_tuned = target_encoder.inverse_transform(y_pred_encoded)

print("Best Parameters:")
print(xgb_search.best_params_)

print("\nBest CV Macro F1:", xgb_search.best_score_)
print("Validation Accuracy:", accuracy_score(y_val, y_pred_xgb_tuned))
print("Validation Macro F1:", f1_score(y_val, y_pred_xgb_tuned, average="macro"))

print("\nValidation Classification Report:")
print(classification_report(y_val, y_pred_xgb_tuned))

```

```python
# =========================
# BMI AS A LABEL-PROXY: SENSITIVITY CHECK
# BMI Only vs Without BMI vs Selected Full Model
# Evaluation is done on the validation set.
# =========================

selected_model_name = final_model_name
selected_model = all_fitted_models[selected_model_name]
selected_estimator = selected_model.named_steps["model"]

print("Selected model used for label-proxy check:", selected_model_name)


# -------------------------
# 1. BMI-Only Model
# -------------------------

X_train_bmi_only = X_train[["BMI"]]
X_val_bmi_only = X_val[["BMI"]]

bmi_only_preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, ["BMI"])
    ],
    remainder="drop"
)

bmi_only_estimator = selected_estimator.__class__(
    **selected_estimator.get_params()
)

bmi_only_model = Pipeline(steps=[
    ("preprocessor", bmi_only_preprocessor),
    ("model", bmi_only_estimator)
])


# -------------------------
# 2. Model Without BMI
# -------------------------

features_without_bmi = [
    feature for feature in final_features
    if feature != "BMI"
]

X_train_without_bmi = X_train[features_without_bmi]
X_val_without_bmi = X_val[features_without_bmi]

numerical_cols_without_bmi = [
    col for col in final_numerical_cols
    if col != "BMI"
]

ordinal_cols_without_bmi = [
    col for col in ordinal_cols
    if col in features_without_bmi
]

nominal_cols_without_bmi = [
    col for col in nominal_cols
    if col in features_without_bmi
]

ordinal_categories_without_bmi = [
    ordinal_categories[ordinal_cols.index(col)]
    for col in ordinal_cols_without_bmi
]

without_bmi_transformers = []

if len(numerical_cols_without_bmi) > 0:
    without_bmi_transformers.append(
        ("num", numeric_transformer, numerical_cols_without_bmi)
    )

if len(ordinal_cols_without_bmi) > 0:
    ordinal_transformer_without_bmi = Pipeline(steps=[
        ("encoder", OrdinalEncoder(
            categories=ordinal_categories_without_bmi,
            handle_unknown="use_encoded_value",
            unknown_value=-1
        ))
    ])

    without_bmi_transformers.append(
        ("ord", ordinal_transformer_without_bmi, ordinal_cols_without_bmi)
    )

if len(nominal_cols_without_bmi) > 0:
    without_bmi_transformers.append(
        ("nom", nominal_transformer, nominal_cols_without_bmi)
    )

preprocessor_without_bmi = ColumnTransformer(
    transformers=without_bmi_transformers,
    remainder="drop"
)

without_bmi_estimator = selected_estimator.__class__(
    **selected_estimator.get_params()
)

model_without_bmi = Pipeline(steps=[
    ("preprocessor", preprocessor_without_bmi),
    ("model", without_bmi_estimator)
])


# -------------------------
# 3. Train and Predict
# Use the same selected model type for all versions.
# -------------------------

if selected_model_name in ["XGBoost", "Tuned XGBoost"]:
    
    bmi_only_model.fit(
        X_train_bmi_only,
        y_train_encoded,
        model__sample_weight=train_sample_weight
    )
    
    bmi_only_pred_encoded = bmi_only_model.predict(X_val_bmi_only)
    y_pred_bmi_only = target_encoder.inverse_transform(bmi_only_pred_encoded)
    
    model_without_bmi.fit(
        X_train_without_bmi,
        y_train_encoded,
        model__sample_weight=train_sample_weight
    )
    
    without_bmi_pred_encoded = model_without_bmi.predict(X_val_without_bmi)
    y_pred_without_bmi = target_encoder.inverse_transform(without_bmi_pred_encoded)

else:
    
    bmi_only_model.fit(
        X_train_bmi_only,
        y_train,
        model__sample_weight=train_sample_weight
    )
    
    y_pred_bmi_only = bmi_only_model.predict(X_val_bmi_only)
    
    model_without_bmi.fit(
        X_train_without_bmi,
        y_train,
        model__sample_weight=train_sample_weight
    )
    
    y_pred_without_bmi = model_without_bmi.predict(X_val_without_bmi)


# -------------------------
# 4. Selected Full Model Prediction
# -------------------------

if selected_model_name in ["XGBoost", "Tuned XGBoost"]:
    full_pred_encoded = selected_model.predict(X_val)
    y_pred_full_selected = target_encoder.inverse_transform(full_pred_encoded)
else:
    y_pred_full_selected = selected_model.predict(X_val)


# -------------------------
# 5. Collect Results
# -------------------------

bmi_only_macro_f1 = f1_score(
    y_val,
    y_pred_bmi_only,
    average="macro"
)

without_bmi_macro_f1 = f1_score(
    y_val,
    y_pred_without_bmi,
    average="macro"
)

full_selected_macro_f1 = f1_score(
    y_val,
    y_pred_full_selected,
    average="macro"
)

drop_without_bmi = full_selected_macro_f1 - without_bmi_macro_f1
gap_full_vs_bmi_only = full_selected_macro_f1 - bmi_only_macro_f1

comparison_bmi_label_proxy = pd.DataFrame({
    "Model Version": [
        "BMI Only",
        "Without BMI",
        f"Selected Full Model ({selected_model_name})"
    ],
    "Accuracy": [
        accuracy_score(y_val, y_pred_bmi_only),
        accuracy_score(y_val, y_pred_without_bmi),
        accuracy_score(y_val, y_pred_full_selected)
    ],
    "Macro F1": [
        bmi_only_macro_f1,
        without_bmi_macro_f1,
        full_selected_macro_f1
    ],
    "Weighted F1": [
        f1_score(y_val, y_pred_bmi_only, average="weighted"),
        f1_score(y_val, y_pred_without_bmi, average="weighted"),
        f1_score(y_val, y_pred_full_selected, average="weighted")
    ]
})

display(comparison_bmi_label_proxy.round(4))

print("BMI as a Label-Proxy Sensitivity Check")
print("BMI Only Macro F1:", round(bmi_only_macro_f1, 4))
print("Without BMI Macro F1:", round(without_bmi_macro_f1, 4))
print("Selected Full Model Macro F1:", round(full_selected_macro_f1, 4))
print("Drop Without BMI:", round(drop_without_bmi, 4))
print("Gap Between Full Model and BMI Only:", round(gap_full_vs_bmi_only, 4))
```

    </div>
    <span style="display:inline-block; padding:5px 10px; border-radius:999px; background:#EAF6FF; color:#0077B6; font-weight:700; font-size:12px; letter-spacing:.4px;">Interpretability</span>
  </div>
  <ul style="margin: 10px 0 0 20px; color:#27464D; line-height:1.65;"><li>Model based importance gives a direct importance estimate where available.</li><li>Permutation importance measures the performance drop when a feature is shuffled.</li><li>Permutation importance is safer when one hot encoded features may inflate categorical importance.</li></ul>
</div>


```python
# =========================
# FEATURE IMPORTANCE
# Model-based Feature Importance
# Note: Summing one-hot encoded columns can inflate categorical feature importance.
# Permutation importance is added later for safer interpretation.
# =========================

def clean_feature_name(feature_name):
    
    feature_name = feature_name.replace("num__", "")
    feature_name = feature_name.replace("ord__", "")
    feature_name = feature_name.replace("nom__", "")
    
    return feature_name


def get_original_feature_name(encoded_feature_name):
    
    for col in final_categorical_cols:
        if encoded_feature_name.startswith(col + "_"):
            return col
    
    return encoded_feature_name


feature_importance_tables = {}

for name, fitted_pipeline in all_fitted_models.items():
    
    model = fitted_pipeline.named_steps["model"]
    
    feature_names = fitted_pipeline.named_steps["preprocessor"].get_feature_names_out()
    feature_names = [clean_feature_name(feature) for feature in feature_names]
    
    if hasattr(model, "feature_importances_"):
        importance_values = model.feature_importances_
        
    elif hasattr(model, "coef_"):
        coef_values = model.coef_
        
        if coef_values.ndim > 1:
            importance_values = np.mean(np.abs(coef_values), axis=0)
        else:
            importance_values = np.abs(coef_values)
    
    else:
        print(f"{name}: Feature importance is not directly available.")
        continue
    
    fi_df = pd.DataFrame({
        "Encoded Feature": feature_names,
        "Importance": importance_values
    })
    
    fi_df["Original Feature"] = fi_df["Encoded Feature"].apply(get_original_feature_name)
    
    fi_agg = (
        fi_df
        .groupby("Original Feature", as_index=False)["Importance"]
        .sum()
        .sort_values("Importance", ascending=False)
    )
    
    fi_agg["Importance"] = (
        fi_agg["Importance"] / fi_agg["Importance"].sum()
    )
    
    feature_importance_tables[name] = fi_agg


print("Feature importance available for:")
print(list(feature_importance_tables.keys()))
```

```python
# =========================
# PERMUTATION IMPORTANCE
# Selected final model
# =========================

best_model_name = final_model_name
best_model = all_fitted_models[best_model_name]


def make_permutation_scorer(model_name):
    
    def scorer(estimator, X_eval, y_eval):
        
        predictions = estimator.predict(X_eval)
        
        # XGBoost models were trained on encoded target labels,
        # so predictions must be converted back to original class names.
        if model_name in ["XGBoost", "Tuned XGBoost"]:
            predictions = target_encoder.inverse_transform(predictions)
        
        return f1_score(y_eval, predictions, average="macro")
    
    return scorer


perm_result = permutation_importance(
    estimator=best_model,
    X=X_val,
    y=y_val,
    scoring=make_permutation_scorer(best_model_name),
    n_repeats=20,
    random_state=42,
    n_jobs=-1
)


permutation_importance_df = pd.DataFrame({
    "Feature": X_val.columns,
    "Mean Importance": perm_result.importances_mean,
    "Std Importance": perm_result.importances_std
})


permutation_importance_df = (
    permutation_importance_df
    .sort_values("Mean Importance", ascending=False)
    .reset_index(drop=True)
)


display(permutation_importance_df.head(15).round(4))


top_perm_features = (
    permutation_importance_df
    .head(15)
    .sort_values("Mean Importance", ascending=True)
)


plt.figure(figsize=(10, 6))

plt.barh(
    top_perm_features["Feature"],
    top_perm_features["Mean Importance"],
    color=HOT_RED,
    edgecolor=DARK_BLUE
)

plt.title(f"Permutation Importance for Selected Final Model: {best_model_name}")
plt.xlabel("Mean Decrease in Macro F1")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
# =========================
# TOP 10 FEATURE SUBSET EXPERIMENT
# Retrain the validation-selected model using only the top 10 features
# selected by validation permutation importance.
# =========================

top10_features = permutation_importance_df.head(10)["Feature"].tolist()

top10_numerical_cols = [
    col for col in final_numerical_cols
    if col in top10_features
]

top10_categorical_cols = [
    col for col in final_categorical_cols
    if col in top10_features
]

top10_ordinal_cols = [
    col for col in ordinal_cols
    if col in top10_features
]

top10_nominal_cols = [
    col for col in top10_categorical_cols
    if col not in top10_ordinal_cols
]

top10_ordinal_categories = [
    ordinal_categories[ordinal_cols.index(col)]
    for col in top10_ordinal_cols
]

print("Top 10 selected features:")
print(top10_features)


# =========================
# TOP 10 DATA SPLIT
# Use the same training and validation sets.
# =========================

X_train_top10 = X_train[top10_features]
X_val_top10 = X_val[top10_features]


# =========================
# TOP 10 PREPROCESSOR
# Built dynamically to handle numerical, ordinal,
# and nominal features only if they are selected.
# =========================

top10_transformers = []

if len(top10_numerical_cols) > 0:
    top10_transformers.append(
        ("num", numeric_transformer, top10_numerical_cols)
    )

if len(top10_ordinal_cols) > 0:
    top10_ordinal_transformer = Pipeline(steps=[
        ("encoder", OrdinalEncoder(
            categories=top10_ordinal_categories,
            handle_unknown="use_encoded_value",
            unknown_value=-1
        ))
    ])

    top10_transformers.append(
        ("ord", top10_ordinal_transformer, top10_ordinal_cols)
    )

if len(top10_nominal_cols) > 0:
    top10_transformers.append(
        ("nom", nominal_transformer, top10_nominal_cols)
    )

top10_preprocessor = ColumnTransformer(
    transformers=top10_transformers
)


# =========================
# REBUILD SAME BEST MODEL TYPE
# =========================

best_model_name = final_model_name
best_model = all_fitted_models[best_model_name]
best_estimator = best_model.named_steps["model"]

top10_estimator = best_estimator.__class__(
    **best_estimator.get_params()
)

top10_model = Pipeline(steps=[
    ("preprocessor", top10_preprocessor),
    ("model", top10_estimator)
])


# =========================
# TRAIN TOP 10 MODEL
# =========================

if best_model_name in ["XGBoost", "Tuned XGBoost"]:
    top10_model.fit(
        X_train_top10,
        y_train_encoded,
        model__sample_weight=train_sample_weight
    )

    top10_pred_encoded = top10_model.predict(X_val_top10)
    y_pred_top10 = target_encoder.inverse_transform(top10_pred_encoded)

else:
    top10_model.fit(
        X_train_top10,
        y_train,
        model__sample_weight=train_sample_weight
    )
    y_pred_top10 = top10_model.predict(X_val_top10)


# =========================
# TOP 10 VALIDATION METRICS
# =========================

top10_accuracy = accuracy_score(y_val, y_pred_top10)

top10_macro_precision = precision_score(
    y_val,
    y_pred_top10,
    average="macro",
    zero_division=0
)

top10_macro_recall = recall_score(
    y_val,
    y_pred_top10,
    average="macro",
    zero_division=0
)

top10_macro_f1 = f1_score(
    y_val,
    y_pred_top10,
    average="macro"
)

top10_weighted_f1 = f1_score(
    y_val,
    y_pred_top10,
    average="weighted"
)

full_best_row = metrics_table.loc[
    metrics_table["Model"] == best_model_name
].iloc[0]

top10_comparison_table = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted F1"
    ],
    "Full Feature Validation Model": [
        full_best_row["Accuracy"],
        full_best_row["Macro Precision"],
        full_best_row["Macro Recall"],
        full_best_row["Macro F1"],
        full_best_row["Weighted F1"]
    ],
    "Top 10 Feature Validation Model": [
        top10_accuracy,
        top10_macro_precision,
        top10_macro_recall,
        top10_macro_f1,
        top10_weighted_f1
    ]
})

top10_comparison_table["Difference"] = (
    top10_comparison_table["Top 10 Feature Validation Model"]
    - top10_comparison_table["Full Feature Validation Model"]
)

display(top10_comparison_table.round(4))

print("Selected final model:", best_model_name)
print("Top 10 feature model Validation Macro F1:", round(top10_macro_f1, 4))
print("Validation Macro F1 difference:", round(top10_macro_f1 - full_best_row["Macro F1"], 4))

if abs(top10_macro_f1 - full_best_row["Macro F1"]) <= 0.005:
    print("Result: The top 10 feature model gives almost the same validation Macro F1 as the full model.")
else:
    print("Result: The top 10 feature model does not give the same validation Macro F1 as the full model.")


# =========================
# VISUAL COMPARISON
# =========================

plot_table = top10_comparison_table.set_index("Metric")[
    ["Full Feature Validation Model", "Top 10 Feature Validation Model"]
]

ax = plot_table.plot(
    kind="bar",
    figsize=(12, 6),
    color=[SHINY_BLUE, SHINY_GREEN],
    edgecolor=DARK_BLUE
)

plt.title("Full Feature Model vs Top 10 Feature Model, Validation Set")
plt.xlabel("Metric")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()


# =========================
# CONFUSION MATRIX
# Top 10 Feature Model on validation set
# =========================

cm_top10 = confusion_matrix(
    y_val,
    y_pred_top10,
    labels=target_encoder.classes_
)

plt.figure(figsize=(7, 5))

plt.imshow(
    cm_top10,
    cmap=HEAT_CMAP
)

plt.title("Top 10 Feature Model Validation Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.xticks(
    ticks=np.arange(len(target_encoder.classes_)),
    labels=target_encoder.classes_,
    rotation=45,
    ha="right"
)

plt.yticks(
    ticks=np.arange(len(target_encoder.classes_)),
    labels=target_encoder.classes_
)

for i in range(cm_top10.shape[0]):
    for j in range(cm_top10.shape[1]):
        plt.text(
            j,
            i,
            cm_top10[i, j],
            ha="center",
            va="center",
            color=DARK_BLUE,
            fontweight="bold"
        )

plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
"""
Obesity Classification — Train on Top 10 Features
================================================

This script trains the XGBoost model on the Top 10 Features selected in Obesity_Last.ipynb:
['BMI', 'CH2O', 'Age', 'TUE', 'FAF', 'NCP', 'FCVC', 'family_history_with_overweight', 'FAVC', 'CALC']
"""

import os
import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    LabelEncoder,
    OrdinalEncoder,
    OneHotEncoder,
    RobustScaler,
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
)
from xgboost import XGBClassifier

# ──────────────────────────────────────────────────────────
# 1. CONFIGURATION
# ──────────────────────────────────────────────────────────

try:
    _BASE_DIR = os.path.dirname(os.path.abspath(__file__))
except NameError:
    _BASE_DIR = os.getcwd()

DATA_PATH = os.path.join(_BASE_DIR, "train.csv")
MODEL_PATH = os.path.join(_BASE_DIR, "obesity_xgb_top10.joblib")

# ──────────────────────────────────────────────────────────
# 2. FEATURE DEFINITIONS (TOP 10 ONLY)
# ──────────────────────────────────────────────────────────

NUMERICAL_COLS = [
    "Age",   
    "FCVC",  
    "NCP",   
    "CH2O",  
    "FAF",   
    "TUE",   
    "BMI",   
]

CATEGORICAL_COLS = [
    "family_history_with_overweight",  
    "FAVC",                           
    "CALC",                           
]

FEATURE_COLS = NUMERICAL_COLS + CATEGORICAL_COLS

ORDINAL_COLS = ["CALC"]
ORDINAL_CATEGORIES = [
    ["no", "Sometimes", "Frequently", "Always"],  # CALC
]

NOMINAL_COLS = [c for c in CATEGORICAL_COLS if c not in ORDINAL_COLS]

TARGET_COL = "NObeyesdad"
TARGET_CLASSES = ["Underweight", "Normal", "Overweight", "Obesity"]

TARGET_GROUP_MAP = {
    "Insufficient_Weight": "Underweight",
    "Normal_Weight": "Normal",
    "Overweight_Level_I": "Overweight",
    "Overweight_Level_II": "Overweight",
    "Obesity_Type_I": "Obesity",
    "Obesity_Type_II": "Obesity",
    "Obesity_Type_III": "Obesity",
}

def build_pipeline():
    numeric_transformer = Pipeline(steps=[
        ("scaler", RobustScaler()),
    ])

    ordinal_transformer = Pipeline(steps=[
        ("encoder", OrdinalEncoder(
            categories=ORDINAL_CATEGORIES,
            handle_unknown="use_encoded_value",
            unknown_value=-1,
        )),
    ])

    nominal_transformer = Pipeline(steps=[
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
    ])

    preprocessor = ColumnTransformer(transformers=[
        ("num", numeric_transformer, NUMERICAL_COLS),
        ("ord", ordinal_transformer, ORDINAL_COLS),
        ("nom", nominal_transformer, NOMINAL_COLS),
    ])

    model = XGBClassifier(
        n_estimators=600,
        max_depth=4,
        learning_rate=0.03,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="multi:softprob",
        eval_metric="mlogloss",
        random_state=42,
        n_jobs=-1,
    )

    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model),
    ])

    return pipeline

def load_and_prepare_data(data_path=DATA_PATH):
    df = pd.read_csv(data_path)

    if "id" in df.columns:
        df = df.drop(columns=["id"])

    df["BMI"] = df["Weight"] / (df["Height"] ** 2)

    if df[TARGET_COL].isin(TARGET_GROUP_MAP.keys()).any():
        df[TARGET_COL] = df[TARGET_COL].map(TARGET_GROUP_MAP)

    X = df[FEATURE_COLS].copy()
    y = df[TARGET_COL].copy()

    return X, y

def compute_sample_weights(y_train_series):
    class_counts = y_train_series.value_counts()
    n_classes = len(class_counts)
    total = len(y_train_series)
    weight_map = {
        cls: total / (n_classes * count)
        for cls, count in class_counts.items()
    }
    return y_train_series.map(weight_map).values

def train_model(evaluate=True, save_path=MODEL_PATH):
    X, y = load_and_prepare_data()

    target_encoder = LabelEncoder()
    target_encoder.classes_ = np.array(TARGET_CLASSES)
    y_encoded = target_encoder.transform(y)

    if evaluate:
        X_train, X_test, y_train, y_test = train_test_split(
            X, y_encoded,
            test_size=0.20,
            random_state=42,
            stratify=y_encoded,
        )

        sample_weights = compute_sample_weights(
            pd.Series(target_encoder.inverse_transform(y_train))
        )

        pipeline = build_pipeline()
        pipeline.fit(
            X_train, y_train,
            model__sample_weight=sample_weights,
        )

        y_pred = pipeline.predict(X_test)
        y_pred_labels = target_encoder.inverse_transform(y_pred)
        y_test_labels = target_encoder.inverse_transform(y_test)

        acc = accuracy_score(y_test_labels, y_pred_labels)
        macro_f1 = f1_score(y_test_labels, y_pred_labels, average="macro")
        weighted_f1 = f1_score(y_test_labels, y_pred_labels, average="weighted")

        print("=" * 55)
        print("  EVALUATION (80/20 split) - Top 10 Features")
        print("=" * 55)
        print(f"  Accuracy     : {acc:.4f}")
        print(f"  Macro F1     : {macro_f1:.4f}")
        print(f"  Weighted F1  : {weighted_f1:.4f}")
        print()
        print(classification_report(
            y_test_labels, y_pred_labels,
            target_names=TARGET_CLASSES,
        ))

    print("Retraining on the full dataset with Top 10 features...")
    sample_weights_full = compute_sample_weights(
        pd.Series(target_encoder.inverse_transform(y_encoded))
    )

    pipeline = build_pipeline()
    pipeline.fit(
        X, y_encoded,
        model__sample_weight=sample_weights_full,
    )

    artifact = {
        "pipeline": pipeline,
        "target_encoder": target_encoder,
        "feature_cols": FEATURE_COLS,
        "numerical_cols": NUMERICAL_COLS,
        "categorical_cols": CATEGORICAL_COLS,
        "target_classes": TARGET_CLASSES,
    }
    joblib.dump(artifact, save_path)
    print(f"\nModel saved to: {save_path}")
    print(f"File size    : {os.path.getsize(save_path) / 1024:.1f} KB")

    return artifact

if __name__ == "__main__":
    artifact = train_model(evaluate=True, save_path=MODEL_PATH)
```

```python
from typing import Iterable, Mapping
from pathlib import Path
import joblib
import numpy as np
import pandas as pd

DEFAULT_BUNDLE_PATH = "obesity_xgb_top10.joblib"

# ---------------------------------------------------------------------------
# Bundle loading
# ---------------------------------------------------------------------------

def load_bundle(bundle_path: str | Path = DEFAULT_BUNDLE_PATH) -> dict:
    """Load the saved joblib bundle and check the keys it must carry."""
    bundle = joblib.load(bundle_path)
    # UPDATED: Matching the exact keys saved by train_top10.py
    required_keys = {
        "pipeline", "feature_cols", "numerical_cols",
        "categorical_cols", "target_classes",
    }
    missing = required_keys - bundle.keys()
    if missing:
        raise ValueError(
            f"Bundle '{bundle_path}' is missing required keys: {sorted(missing)}"
        )
    return bundle

# ---------------------------------------------------------------------------
# Input preparation
# ---------------------------------------------------------------------------

def _ensure_bmi(df: pd.DataFrame, bundle: dict) -> pd.DataFrame:
    """If the model needs BMI but the user supplied only Height + Weight,
    compute BMI from them."""
    if "BMI" not in bundle["feature_cols"]: # UPDATED
        return df
    if "BMI" in df.columns and df["BMI"].notna().all():
        return df
    if {"Height", "Weight"}.issubset(df.columns):
        df = df.copy()
        df["BMI"] = df["Weight"] / (df["Height"] ** 2)
        return df
    raise ValueError(
        "Model expects a BMI column. Either pass BMI directly, "
        "or pass both Height (m) and Weight (kg) so it can be computed."
    )

def prepare_input(
    data: pd.DataFrame | Mapping | Iterable[Mapping],
    bundle: dict,
) -> pd.DataFrame:
    if isinstance(data, pd.DataFrame):
        df = data.copy()
    elif isinstance(data, Mapping):
        df = pd.DataFrame([dict(data)])
    else:
        df = pd.DataFrame(list(data))

    df = _ensure_bmi(df, bundle)

    expected = bundle["feature_cols"] # UPDATED
    missing = [c for c in expected if c not in df.columns]
    if missing:
        raise ValueError(
            f"Missing required input columns: {missing}. "
            f"Expected (in order): {expected}"
        )

    return df[expected]

# ---------------------------------------------------------------------------
# Prediction
# ---------------------------------------------------------------------------

def _decode(predictions: np.ndarray, bundle: dict) -> np.ndarray:
    """Decode back to original class names."""
    if "target_encoder" in bundle and bundle["target_encoder"] is not None:
        return bundle["target_encoder"].inverse_transform(predictions)
    return predictions

def predict(
    data: pd.DataFrame | Mapping | Iterable[Mapping],
    bundle_path: str | Path = DEFAULT_BUNDLE_PATH,
    return_proba: bool = False,
) -> pd.DataFrame:
    bundle = load_bundle(bundle_path)
    X = prepare_input(data, bundle)
    pipe = bundle["pipeline"] # UPDATED

    preds = _decode(pipe.predict(X), bundle)
    out = pd.DataFrame({"prediction": preds}, index=X.index)

    if return_proba and hasattr(pipe, "predict_proba"):
        proba = pipe.predict_proba(X)
        if "target_encoder" in bundle and bundle["target_encoder"] is not None:
            class_names = bundle["target_encoder"].classes_
        else:
            class_names = pipe.classes_
        proba_df = pd.DataFrame(
            proba,
            columns=[f"proba_{c}" for c in class_names],
            index=X.index,
        )
        out = pd.concat([out, proba_df], axis=1)

    return out

def predict_one(
    patient: Mapping,
    bundle_path: str | Path = DEFAULT_BUNDLE_PATH,
    return_proba: bool = True,
) -> dict:
    return predict(patient, bundle_path=bundle_path, return_proba=return_proba).iloc[0].to_dict()


# ===========================================================================
# PREDICTION TEST
# ===========================================================================

patient_data = {
    # "Gender": "Female",
    "Age": 38,
    "Height": 1.85,   
    "Weight": 130.0,   
    "family_history_with_overweight": "yes", # yes / no
    "FCVC": 2.0,
    "NCP": 3.0,
    "CH2O": 2.0,
    "FAF": 1.0,
    "TUE": 1.0,
    "CALC": "Sometimes", # Sometimes / no / Frequently
    "FAVC": "yes"        # yes / no
}

result = predict_one(patient_data)
print("Single Patient Prediction:")
print(f"  Predicted Class: {result['prediction']}")
```
